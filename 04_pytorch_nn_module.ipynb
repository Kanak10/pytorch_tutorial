```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features)
model(features)
```

```python
# show model weights
model.linear.weight
```

```python
# show model bias
model.linear.bias
```

## Complex Neural Network

```python
# create model class
import torch
import torch.nn as nn

class ComplexModel(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear1(features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
complex_model = ComplexModel(features.shape[1])

# call model for forward pass
# complex_model.forward(features)
complex_model(features)
```

```python
complex_model.linear1.weight, complex_model.linear2.weight
```

```python
complex_model.linear1.bias, complex_model.linear2.bias
```

```python
summary(complex_model, input_size=(10, 5))
```

## Sequantial Container

```python
# create model class
import torch
import torch.nn as nn

class SequentialContainerModel(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
      nn.Linear(num_features, 3),
      nn.ReLU(),
      nn.Linear(3, 1),
      nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
sequential_container_model = SequentialContainerModel(features.shape[1])

# call model for forward pass
# sequential_container_model.forward(features)
sequential_container_model(features)
```
